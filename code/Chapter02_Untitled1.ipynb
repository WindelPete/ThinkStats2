```python
import nsfg2
```

```python
preg = nsfg2.ReadFemPreg()
live = preg[preg.outcome ==1]
```

```python
live
```

```python
hist = thinkstats2.Hist(live.birthwgt_lb1, label='birthwgt_lb1')
thinkplot.Hist(hist)
thinkplot.Show(xlabel='pounds', ylabel='frequency')
```

```python
hist = thinkstats2.Hist(live.agepreg, label='agepreg')
thinkplot.Hist(hist)
thinkplot.Show(xlabel='years', ylabel='frequency')
```

```python
hist = thinkstats2.Hist(live.prglngth, label='agepreg')
thinkplot.Hist(hist)
thinkplot.Show(xlabel='Weeks', ylabel='frequency')
```

```python
hist = thinkstats2.Hist(live.prglngth, label='birthwgt_lb1')
for weeks, freq in hist.Smallest(10):
    print(weeks, freq)
```

```python
hist = thinkstats2.Hist(live.prglngth, label='birthwgt_lb1')
for weeks, freq in hist.Largest(10):
    print(weeks, freq)
```

```python
firsts = live[live.birthord == 1]
others = live[live.birthord != 1]
first_hist = thinkstats2.Hist(firsts.prglngth, label='first')
other_hist = thinkstats2.Hist(others.prglngth, label='other')
width = 0.45
thinkplot.PrePlot(2)
thinkplot.Hist(first_hist, align='right', width=width)
thinkplot.Hist(other_hist, align='left', width=width)
thinkplot.Show(xlabel='weeks', ylabel='frequency',
xlim=[27, 46])
```

```python
mean = live.prglngth.mean()
var = live.prglngth.var()
std = live.prglngth.std()
print(mean, var, std)
```
